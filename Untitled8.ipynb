```r
stimulateBirthdayParties <- function(n, simulations=1000){
  matchCount=0
  for(i in 1:simulations){
    birthdays=sample(1:365,n,replace=TRUE)
    # print(birthdays)
    # print(unique(birthdays))
    if(length(unique(birthdays))<n){
      matchCount=matchCount+1
    }
  }

  prob = matchCount/simulations
  return(prob)
}

ns=5:50
prob = sapply(ns,stimulateBirthdayParties)
print(prob)
```

```r
calculate_birthday_probability <- function(n) {
  prod <- 1
  for (i in 1:(n-1)) {
    prod <- prod * (1 - i / 365)
  }
  probability <- 1 - prod
  return(probability)
}

n <- 20
while (calculate_birthday_probability(n) <= 0.5) {
  n <- n + 1
}
cat("The smallest n where probability exceeds 0.5 is:", n, "\n")
```

```r
calculate_mode <- function(x) {
  freq_table <- table(x)
  max_freq <- max(freq_table)
  print(freq_table)

  mode_values <- as.numeric(names(freq_table[freq_table == max_freq]))

  # If there's a single mode, return it. Otherwise, return all modes.
  if(length(mode_values) == 1) {
    return(mode_values)
  } else {
    return(mode_values)
  }
}

data2 <- c(1, 2, 2, 3, 4, 4, 4, 5, 6, 6, 6)
mode_value <- calculate_mode(data2)
paste("Mode:", mode_value)
```

```r
# Define probability function for X (Geometric distribution)
p_x <- function(x) {
  (3/4) * ((1/4)^(x-1))
}

# Define function for probability of Y = X^2
p_y <- function(y) {
  x <- sqrt(y)  # Since Y = X^2 => X = sqrt(Y)
  if (x == floor(x)) {  # Ensure X is an integer
    return(p_x(x))
  } else {
    return(0)
  }
}

# Compute probability of Y for X = 3 (i.e., Y = 3^2 = 9)
p_Y_9 <- p_y(9)
print(paste("P(Y=9):", p_Y_9))

# Define X values
X_vals <- 1:5
Y_vals <- X_vals^2

# Compute probability distribution of Y
P_Y <- sapply(Y_vals, p_y)

# Compute expected value (E[Y])
E_Y <- weighted.mean(Y_vals, P_Y)

# Compute second moment E[Y^2]
E_Y2 <- weighted.mean(Y_vals^2, P_Y)

# Compute variance Var(Y) = E[Y^2] - (E[Y])^2
Var_Y <- E_Y2 - (E_Y)^2

# Print results
print(paste("Expected Value E[Y]:", E_Y))
print(paste("Second Moment E[Y^2]:", E_Y2))
print(paste("Variance Var(Y):", Var_Y))
```

```r
#6.2
# The joint probability mass function of two random variables X and Y is
# f(x, y) = (x+y)/30; x = 0,1,2,3; y = 0,1,2
# Write a R-code to
# (i) display the joint mass function in rectangular (matrix) form.
# (ii) check that it is joint mass function or not? (use: Sum())
# (iii) find the marginal distribution g(x) for x = 0,1,2,3. (Use:apply())
# (iv) find the marginal distribution h(y) for y = 0,1,2. (Use:apply())
# (v) find the conditional probability at x = 0 given y = 1.
# (vi) find E(x), E(y), E(xy), Var(x), Var(y), Cov(x, y) and its correlation coefficient.

func = function(x,y){
  return ((x*y)/30)
}

Xvals = c(0:3)
Yvals = c(0:2)

#2.1
m = matrix(c(
  func(Xvals[1],Yvals), func(Xvals[2],Yvals), func(Xvals[3],Yvals), func(Xvals[4],Yvals)
), nrow =4, ncol=3, byrow = TRUE)
print(m)

#2.2
sum = sum(m)
print(sum)
if(sum == 1){
  print("PDF")
}

#2.3 and 2.4
gX = apply(m,1,sum) # 1 means row wise operation, gives sum row wise
hY = apply(m,2,sum) # 2 means col wise operation, gives sum col wise
print(gX)
print(hY)

#2.5
print(m[1,2]/hY)

#2.6
eX = sum(Xvals*gX)
eY = sum(Yvals*hY)

func2 = function(x,y){
  return(x*y*func(x,y))
}

m2 = matrix(c(
  func2(Xvals[1],Yvals), func2(Xvals[2],Yvals), func2(Xvals[3],Yvals), func2(Xvals[4],Yvals)
), nrow =4, ncol=3, byrow = TRUE)

eXY = sum(m2)
varX = sum(((Xvals - eX)^2)*gX)
varY = sum(((Yvals - eY)^2)*hY)
covXY = eXY - eX*eY
cor = covXY/sqrt(varX * varY)

cat("E(X):", eX, "\n")
cat("E(Y):", eY, "\n")
cat("E(XY):", eXY, "\n")
cat("Var(X):", varX, "\n")
cat("Var(Y):", varY, "\n")
cat("Cov(X, Y):", covXY, "\n")
cat("Correlation Coefficient:", cor, "\n")
```

```r
set.seed(123)
#8.2
# For X ∼ Exponential(λ = 1.5), simulate sample means and check normality for
#   n = 10 and n = 50.
# 2. Does the Central Limit Theorem apply here? How does sample size affect the
#   result?

#1.1
lambda = 1.5
N = 1000

stimulatedMeans = function(n){
  replicate(N,mean(rexp(n,rate=lambda)))
}

means10 = stimulatedMeans(10)
means50 = stimulatedMeans(50)

par(mfrow = c(2, 2))

# qqnorm() is a function in R that creates a Q-Q (quantile-quantile) plot
#   to visually check if a set of data follows a normal distribution.
# qqline(data): Adds a reference line (theoretical normal quantiles) to compare with your data.

hist(means10, main = "Sample Means (n=10)", col = "lightblue", breaks = 30)
qqnorm(means10); qqline(means10, col = "red")

hist(means50, main = "Sample Means (n=50)", col = "lightgreen", breaks = 30)
qqnorm(means50); qqline(means50, col = "red")

# But as per CLT, the distribution of the sample mean will:
# Approach a normal distribution as sample size n increases
# Have mean = 1/λ = 0.6667
# Have variance = (1/λ²)/n
```
